```python
import pandas as pd

data = pd.DataFrame({
    'emp_id': [101,102,103,104,105,106,107,108,109,110],
    'name': ['Arjun','Ravi','Priya','Amit','Neha','Kiran','Sneha','Raj','Pooja','Vikas'],
    'department': ['Sales','sales','HR','hr','IT','it','Software','software','Support','support'],
    'salary': [50000,60000,None,45000,70000,None,80000,75000,40000,None],
    'experience': [2,5,3,1,6,4,7,5,2,3],
    'city': ['Mumbai','Pune','Mumbai','Delhi','Pune','Mumbai','Delhi','Pune','Mumbai','Delhi']
})

data
```

```python
#🧹 TASK 1: Data Cleaning
```

```python
data['department']=data['department'].str.replace(r'(?i)sales','Sales',regex=True)
data['department']=data['department'].str.replace(r'(?i)hr','HR',regex=True)
data['department']=data['department'].str.replace(r'(?i)support','Support',regex=True)
data['department']=data['department'].str.replace(r'(?i)IT','IT',regex=True)
data['department']=data['department'].str.replace(r'(?i)Software','Software',regex=True)
data
```

```python
#Fill missing salary with average salary
```

```python

data['salary']=data['salary'].fillna(data['salary'].mean())
```
